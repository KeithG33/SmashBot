# Dataset Extraction from Slippi Files

##### Extract Dataset

```python
from smashbot.data.extract_action_dataset import extract_dataset

# Path to directory containing the Slippi files
SLIPPI_DIR = '/home/kage/smashbot_workspace/dataset/Slippi_Public_Dataset_v3/slp_test'
OUTPUT_DIR = '/home/kage/smashbot_workspace/dataset/hickle_action'
num_workers = 1
chunk_size = 85

extract_dataset(SLIPPI_DIR, OUTPUT_DIR, num_workers, chunk_size)
```

##### Shuffle the dataset

```python
from smashbot.data.shuffle import shuffle_files

# Check the files produced from extract_dataset for the maximum sequence
# length S, hickle files are numpy arrays of shape (B,S,21)
shuffle_output_dir = '/home/kage/smashbot_workspace/dataset/SlippiGames/hickle_shuffle'
for id in range(3, 0):
    print(f"Starting shuffle for id {id}")
    shuffle_files(
        OUTPUT_DIR,
        '/home/kage/smashbot_workspace/dataset/Slippi_Public_Dataset_v3/hickle_shuffle2',
        10,
        num_processes=2,
        num_save_files=8,
        id=id
    )
```
